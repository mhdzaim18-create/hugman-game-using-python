```python
import random

def play_hangman():
    # Predefined list of 5 words as per task instructions 
    words = ["python", "software", "alpha", "script", "coding"]
    
    # Select a random word from the list 
    target_word = random.choice(words)
    guessed_letters = []
    incorrect_guesses = 0
    max_attempts = 6 # Limit set by task scope 

    print("--- Welcome to Zaim Hangman! ---")
    
    while incorrect_guesses < max_attempts:
        # Display current progress 
        display_word = ""
        for letter in target_word:
            if letter in guessed_letters:
                display_word += letter + " "
            else:
                display_word += "_ "
        
        print(f"\nWord: {display_word}")
        print(f"Attempts remaining: {max_attempts - incorrect_guesses}")
        print(f"Guessed letters: {', '.join(guessed_letters)}")

        # Check for win condition
        if "_" not in display_word:
            print(f"\nCongratulations! You guessed the word: {target_word}")
            break

        # Get user input 
        guess = input("Guess a letter: ").lower()

        # Input validation
        if len(guess) != 1 or not guess.isalpha():
            print("Please enter a single valid letter.")
            continue
        
        if guess in guessed_letters:
            print(f"You already guessed '{guess}'. Try a different letter.")
            continue

        guessed_letters.append(guess)

        # Check if guess is correct 
        if guess in target_word:
            print(f"Good job! '{guess}' is in the word.")
        else:
            incorrect_guesses += 1
            print(f"Sorry, '{guess}' is not there.")

    # Check for loss condition
    if incorrect_guesses == max_attempts:
        print(f"\nGame Over! You ran out of attempts. The word was: {target_word}")

if __name__ == "__main__":
    play_hangman()
```
